Categorize the files as positive negative or neutral reviews. This information should be stored somewhere incase we need to access it later. Count the total number of reviews in each category and then print to the screen the total number of positive negative and neutral reviews as a complete sentence. 

You do not need to use anything that I have provided here for you if you have your own ideas but these could be useful

```python
from pathlib import Path
import json
import re

from helper_functions import get_llm_response

CATEGORIES = {"positive", "negative", "neutral"}


def classify_review(text, filename):
    prompt = (
        "You are a sentiment analysis assistant. Read the product review below and "
        "classify its overall sentiment as positive, negative, or neutral. "
        "Reply with exactly one of these words and nothing else.\n\n"
        f"Review from {filename}:\n{text}"
    )
    response = get_llm_response(prompt)
    response = response.strip().lower()
    match = re.fullmatch(r"(?:the review is )?(positive|negative|neutral)[.!]?", response)
    if match is None:
        raise ValueError(f"AI returned an invalid category for {filename}: {response!r}")
    return match.group(1)


def cat_text_files():
    directories = (Path("data"), Path("../data"), Path("."))
    directory = next(
        (candidate for candidate in directories if any(candidate.glob("coffeepotreview*.txt"))),
        None,
    )
    if directory is None:
        raise FileNotFoundError("No coffeepotreview*.txt files found in data/, ../data/, or the current directory.")

    review_files = sorted(directory.glob("coffeepotreview*.txt"))
    categorized = {}
    for review_file in review_files:
        review_text = review_file.read_text(encoding="utf-8").strip()
        category = classify_review(review_text, review_file.name)
        categorized[review_file.name] = category

    output_path = directory / "categorized_reviews.json"
    output_path.write_text(json.dumps(categorized, indent=2), encoding="utf-8")
    return categorized


categorized = cat_text_files()
```

```python
import pprint #This is a function called pretty print to print out things in a easily readable format
pprint.pprint(categorized)
```

```python
def count_value(category, reviews=categorized):
    if category not in CATEGORIES:
        raise ValueError(f"Unknown review category: {category!r}")
    return sum(review_category == category for review_category in reviews.values())
```

```python
positive = count_value("positive")
negative = count_value("negative")
neutral = count_value("neutral")
print(
    f"There are {positive} positive reviews, {negative} negative reviews, "
    f"and {neutral} neutral reviews."
)

```
